```python
import os
import pandas as pd
import numpy as np
from pathlib import Path

print("Kaggle inputs:\n")

all_files = []

for dirname, _, filenames in os.walk("/kaggle/input"):
    for filename in filenames:
        path = os.path.join(dirname, filename)
        all_files.append(path)
        print(path)

print("\nTotal files:", len(all_files))
```

```python
tabular_files = []

for path in all_files:
    if path.lower().endswith((".csv", ".xlsx", ".xls", ".parquet")):
        tabular_files.append(path)

print("Total tabular files:", len(tabular_files))
print("=" * 100)

file_info = []

for i, path in enumerate(tabular_files):
    print(f"\nFILE {i}")
    print(path)

    try:
        if path.lower().endswith(".csv"):
            temp_df = pd.read_csv(path)

        elif path.lower().endswith((".xlsx", ".xls")):
            temp_df = pd.read_excel(path)

        elif path.lower().endswith(".parquet"):
            temp_df = pd.read_parquet(path)

        print("Shape:", temp_df.shape)
        print("Columns:")
        print(temp_df.columns.tolist())

        file_info.append({
            "index": i,
            "path": path,
            "rows": temp_df.shape[0],
            "columns": temp_df.shape[1],
            "column_names": temp_df.columns.tolist()
        })

    except Exception as e:
        print("ERROR:", e)

    print("-" * 100)
```

```python
required_keywords = {
    "temperature": [
        "temp",
        "temperature",
        "air_temp",
        "indoor_temp"
    ],

    "humidity": [
        "humidity",
        "rh",
        "relative_humidity"
    ],

    "soil_moisture": [
        "soil_moisture",
        "soil moisture",
        "moisture",
        "water_content",
        "water content",
        "vwc",
        "substrate_water",
        "substrate moisture"
    ],

    "light": [
        "light",
        "radiation",
        "solar",
        "par"
    ],

    "co2": [
        "co2",
        "carbon_dioxide",
        "carbon dioxide"
    ],

    "soil_temperature": [
        "soil_temp",
        "soil temperature",
        "substrate_temp"
    ]
}


def find_matching_columns(columns, keywords):
    matches = []

    for col in columns:
        col_lower = str(col).lower()

        for keyword in keywords:
            if keyword.lower() in col_lower:
                matches.append(col)
                break

    return list(dict.fromkeys(matches))


print("=" * 100)
print("REQUIRED GREENHOUSE FEATURE SEARCH")
print("=" * 100)

for info in file_info:

    print(f"\nFILE {info['index']}:")
    print(info["path"])

    for feature, keywords in required_keywords.items():

        matches = find_matching_columns(
            info["column_names"],
            keywords
        )

        symbol = "✅" if matches else "❌"

        print(
            f"{symbol} {feature:20s}: "
            f"{matches if matches else 'NOT FOUND'}"
        )

    print("-" * 100)
```

```python
candidate_rows = []

for info in file_info:

    result = {
        "file_index": info["index"],
        "file": Path(info["path"]).name,
        "rows": info["rows"]
    }

    score = 0

    for feature, keywords in required_keywords.items():

        matches = find_matching_columns(
            info["column_names"],
            keywords
        )

        result[feature] = ", ".join(matches)

        if matches:
            score += 1

    result["feature_score"] = score

    candidate_rows.append(result)

candidate_df = pd.DataFrame(candidate_rows)

candidate_df = candidate_df.sort_values(
    "feature_score",
    ascending=False
)

display(candidate_df)
```

```python
# ============================================================
# MERGE GREENHOUSE CLIMATE + ROOT-ZONE SENSOR DATA
# ============================================================

import pandas as pd
import numpy as np
import os

base = "/kaggle/input/datasets/piantic/autonomous-greenhouse-challengeagc-2nd-2019/TheAutomators"

climate_path = os.path.join(base, "GreenhouseClimate.csv")
grodan_path = os.path.join(base, "GrodanSens.csv")

# Read without low-memory inference
climate = pd.read_csv(climate_path, low_memory=False)
grodan = pd.read_csv(grodan_path, low_memory=False)

# ------------------------------------------------------------
# 1. Keep only the variables we need
# ------------------------------------------------------------

climate_cols = [
    "%time",
    "Tair",
    "Rhair",
    "CO2air",
    "Tot_PAR"
]

grodan_cols = [
    "%time",
    "WC_slab1",
    "WC_slab2",
    "t_slab1",
    "t_slab2",
    "EC_slab1",
    "EC_slab2"
]

climate_small = climate[climate_cols].copy()
grodan_small = grodan[grodan_cols].copy()

# ------------------------------------------------------------
# 2. Convert sensor columns to numeric
# ------------------------------------------------------------

for col in climate_cols[1:]:
    climate_small[col] = pd.to_numeric(
        climate_small[col],
        errors="coerce"
    )

for col in grodan_cols[1:]:
    grodan_small[col] = pd.to_numeric(
        grodan_small[col],
        errors="coerce"
    )

# ------------------------------------------------------------
# 3. Merge using the common experiment timestamp
# ------------------------------------------------------------

merged = pd.merge(
    climate_small,
    grodan_small,
    on="%time",
    how="inner",
    validate="one_to_one"
)

# ------------------------------------------------------------
# 4. Convert Excel serial timestamp to real datetime
# ------------------------------------------------------------

merged["datetime"] = pd.to_datetime(
    merged["%time"],
    unit="D",
    origin="1899-12-30"
)

# ------------------------------------------------------------
# 5. Create averaged root-zone variables
# ------------------------------------------------------------

merged["soil_moisture"] = merged[
    ["WC_slab1", "WC_slab2"]
].mean(axis=1)

merged["soil_temperature"] = merged[
    ["t_slab1", "t_slab2"]
].mean(axis=1)

merged["rootzone_ec"] = merged[
    ["EC_slab1", "EC_slab2"]
].mean(axis=1)

# ------------------------------------------------------------
# 6. Rename main farmer-facing variables
# ------------------------------------------------------------

merged = merged.rename(columns={
    "Tair": "temperature",
    "Rhair": "humidity",
    "CO2air": "co2",
    "Tot_PAR": "light_intensity"
})

# ------------------------------------------------------------
# 7. Show availability
# ------------------------------------------------------------

main_features = [
    "temperature",
    "humidity",
    "soil_moisture",
    "light_intensity",
    "co2",
    "soil_temperature",
    "rootzone_ec"
]

print("=" * 85)
print("MERGED GREENHOUSE DATA")
print("=" * 85)

print("Merged rows:", len(merged))

print(
    "Date range:",
    merged["datetime"].min(),
    "to",
    merged["datetime"].max()
)

print("\nMISSING VALUES:")
print(
    merged[main_features]
    .isnull()
    .sum()
)

print("\nMISSING PERCENTAGE:")
print(
    (
        merged[main_features]
        .isnull()
        .mean() * 100
    ).round(2)
)

print("\nVALID VALUES:")
print(
    merged[main_features]
    .notnull()
    .sum()
)

print("\nSENSOR RANGES:")
display(
    merged[main_features]
    .describe()
    .T
    .round(3)
)

print("\nFIRST ROWS WHERE SOIL MOISTURE EXISTS:")

display(
    merged[
        merged["soil_moisture"].notna()
    ][
        [
            "datetime",
            "temperature",
            "humidity",
            "soil_moisture",
            "light_intensity",
            "co2",
            "soil_temperature",
            "rootzone_ec"
        ]
    ].head(10)
)
```

```python
# ============================================================
# NAMMA KRUSHI - COMBINE ALL AGC TOMATO GREENHOUSE COMPARTMENTS
# ============================================================

import os
import pandas as pd
import numpy as np
from pathlib import Path

DATASET_ROOT = (
    "/kaggle/input/datasets/piantic/"
    "autonomous-greenhouse-challengeagc-2nd-2019"
)

TEAMS = [
    "TheAutomators",
    "Automatoes",
    "IUACAAS",
    "Reference",
    "AICU",
    "Digilog"
]

all_team_data = []

for team in TEAMS:

    print(f"\nProcessing: {team}")

    climate_path = os.path.join(
        DATASET_ROOT,
        team,
        "GreenhouseClimate.csv"
    )

    grodan_path = os.path.join(
        DATASET_ROOT,
        team,
        "GrodanSens.csv"
    )

    climate = pd.read_csv(
        climate_path,
        low_memory=False
    )

    grodan = pd.read_csv(
        grodan_path,
        low_memory=False
    )

    # --------------------------------------------------------
    # Required climate features
    # --------------------------------------------------------

    climate_small = climate[
        [
            "%time",
            "Tair",
            "Rhair",
            "CO2air",
            "Tot_PAR"
        ]
    ].copy()

    # --------------------------------------------------------
    # Root-zone sensor features
    # --------------------------------------------------------

    grodan_small = grodan[
        [
            "%time",
            "WC_slab1",
            "WC_slab2",
            "t_slab1",
            "t_slab2",
            "EC_slab1",
            "EC_slab2"
        ]
    ].copy()

    # Convert sensor columns to numeric
    for col in climate_small.columns[1:]:
        climate_small[col] = pd.to_numeric(
            climate_small[col],
            errors="coerce"
        )

    for col in grodan_small.columns[1:]:
        grodan_small[col] = pd.to_numeric(
            grodan_small[col],
            errors="coerce"
        )

    # --------------------------------------------------------
    # Merge climate + root-zone measurements
    # --------------------------------------------------------

    team_df = pd.merge(
        climate_small,
        grodan_small,
        on="%time",
        how="inner",
        validate="one_to_one"
    )

    # --------------------------------------------------------
    # Convert timestamp
    # --------------------------------------------------------

    team_df["datetime"] = pd.to_datetime(
        team_df["%time"],
        unit="D",
        origin="1899-12-30"
    )

    # --------------------------------------------------------
    # Average duplicate root-zone sensors
    # --------------------------------------------------------

    team_df["soil_moisture"] = team_df[
        ["WC_slab1", "WC_slab2"]
    ].mean(axis=1)

    team_df["soil_temperature"] = team_df[
        ["t_slab1", "t_slab2"]
    ].mean(axis=1)

    team_df["rootzone_ec"] = team_df[
        ["EC_slab1", "EC_slab2"]
    ].mean(axis=1)

    # --------------------------------------------------------
    # Rename farmer-facing variables
    # --------------------------------------------------------

    team_df = team_df.rename(
        columns={
            "Tair": "temperature",
            "Rhair": "humidity",
            "CO2air": "co2",
            "Tot_PAR": "light_intensity"
        }
    )

    team_df["team"] = team

    all_team_data.append(team_df)

    print(
        "Rows:",
        len(team_df),
        "| Valid soil moisture:",
        team_df["soil_moisture"].notna().sum()
    )


# ============================================================
# CONCATENATE ALL GREENHOUSE COMPARTMENTS
# ============================================================

greenhouse_all = pd.concat(
    all_team_data,
    ignore_index=True
)

MAIN_FEATURES = [
    "temperature",
    "humidity",
    "soil_moisture",
    "light_intensity",
    "co2"
]

OPTIONAL_FEATURES = [
    "soil_temperature",
    "rootzone_ec"
]

print("\n" + "=" * 90)
print("ALL GREENHOUSE COMPARTMENTS COMBINED")
print("=" * 90)

print("Total rows:", len(greenhouse_all))

print("\nRows per team:")
print(
    greenhouse_all["team"]
    .value_counts()
)

print("\nMissing percentage by feature:")

print(
    greenhouse_all[
        MAIN_FEATURES + OPTIONAL_FEATURES
    ]
    .isnull()
    .mean()
    .mul(100)
    .round(2)
)


# ============================================================
# CREATE FINAL CLEAN DATASET
# ============================================================

# Keep only rows where ALL 5 farmer-facing parameters exist.

greenhouse_clean = greenhouse_all.dropna(
    subset=MAIN_FEATURES
).copy()

greenhouse_clean = greenhouse_clean.sort_values(
    ["team", "datetime"]
).reset_index(drop=True)

print("\n" + "=" * 90)
print("FINAL CLEAN GREENHOUSE DATA")
print("=" * 90)

print(
    "Rows before cleaning:",
    len(greenhouse_all)
)

print(
    "Rows after requiring all 5 main features:",
    len(greenhouse_clean)
)

print(
    "Rows removed:",
    len(greenhouse_all) - len(greenhouse_clean)
)

print("\nFinal sensor ranges:")

display(
    greenhouse_clean[
        MAIN_FEATURES + OPTIONAL_FEATURES
    ]
    .describe()
    .T
    .round(3)
)


# ============================================================
# TEAM-WISE RANGES
# ============================================================

print("\nTEAM-WISE MEANS:")

display(
    greenhouse_clean.groupby("team")[
        MAIN_FEATURES + OPTIONAL_FEATURES
    ]
    .mean()
    .round(2)
)


# ============================================================
# SAVE THIS PERMANENTLY
# ============================================================

output_dir = Path(
    "/kaggle/working/namma_krushi_greenhouse"
)

output_dir.mkdir(
    parents=True,
    exist_ok=True
)

greenhouse_all.to_csv(
    output_dir / "greenhouse_all_teams_raw_merged.csv",
    index=False
)

greenhouse_clean.to_csv(
    output_dir / "greenhouse_all_teams_clean.csv",
    index=False
)

print("\nSaved:")
print(
    output_dir /
    "greenhouse_all_teams_raw_merged.csv"
)

print(
    output_dir /
    "greenhouse_all_teams_clean.csv"
)
```

```python
# ============================================================
# NAMMA KRUSHI - FINAL CLEANING + TARGET CREATION
# ============================================================

import pandas as pd
import numpy as np
import json
from pathlib import Path

final_df = greenhouse_clean.copy()

# ============================================================
# 1. REMOVE ONLY CLEAR SENSOR ERRORS
# ============================================================

valid_mask = (
    (final_df["temperature"] > 0) &
    (final_df["temperature"] <= 45) &

    (final_df["humidity"] >= 0) &
    (final_df["humidity"] <= 100) &

    (final_df["soil_moisture"] > 0) &
    (final_df["soil_moisture"] <= 100) &

    (final_df["light_intensity"] >= 0) &
    (final_df["light_intensity"] <= 2000) &

    (final_df["co2"] >= 250) &
    (final_df["co2"] <= 2500)
)

before = len(final_df)

final_df = (
    final_df.loc[valid_mask]
    .copy()
    .reset_index(drop=True)
)

after = len(final_df)

print("Rows before sensor cleaning:", before)
print("Rows after sensor cleaning :", after)
print("Rows removed              :", before - after)


# ============================================================
# 2. STATUS FUNCTION
# ============================================================
#
# status:
# 0 = Suitable range
# 1 = Needs adjustment / marginal range
# 2 = Unsuitable / severe range
#

def range_status(
    value,
    suitable_low,
    suitable_high,
    marginal_low,
    marginal_high
):
    if suitable_low <= value <= suitable_high:
        return 0

    elif marginal_low <= value <= marginal_high:
        return 1

    else:
        return 2


# ============================================================
# 3. TEMPERATURE
# ============================================================
#
# Practical greenhouse-tomato project ranges:
#
# Suitable: 18–27 C
# Marginal: 15–32 C
#

final_df["temperature_status"] = final_df["temperature"].apply(
    lambda x: range_status(
        x,
        suitable_low=18,
        suitable_high=27,
        marginal_low=15,
        marginal_high=32
    )
)


# ============================================================
# 4. RELATIVE HUMIDITY
# ============================================================
#
# Suitable: 65–90 %
# Marginal: 50–94 %
#

final_df["humidity_status"] = final_df["humidity"].apply(
    lambda x: range_status(
        x,
        suitable_low=65,
        suitable_high=90,
        marginal_low=50,
        marginal_high=94
    )
)


# ============================================================
# 5. ROOT-ZONE / SUBSTRATE MOISTURE
# ============================================================
#
# IMPORTANT:
# WC_slab represents rockwool/substrate water content.
#
# Dataset itself shows most successful operating values
# approximately around the 70–85% region.
#
# Suitable: 68–88 %
# Marginal: 60–95 %
#

final_df["soil_moisture_status"] = final_df["soil_moisture"].apply(
    lambda x: range_status(
        x,
        suitable_low=68,
        suitable_high=88,
        marginal_low=60,
        marginal_high=95
    )
)


# ============================================================
# 6. CO2
# ============================================================
#
# Ambient greenhouse CO2 can be around 400 ppm.
# Enriched greenhouse operation commonly runs higher.
#
# Suitable: 400–1200 ppm
# Marginal: 300–1500 ppm
#

final_df["co2_status"] = final_df["co2"].apply(
    lambda x: range_status(
        x,
        suitable_low=400,
        suitable_high=1200,
        marginal_low=300,
        marginal_high=1500
    )
)


# ============================================================
# 7. LIGHT / PAR
# ============================================================
#
# Tot_PAR is instantaneous PAR.
#
# IMPORTANT:
# 0 PAR must NOT automatically mean "bad" because nighttime
# naturally has zero light.
#
# Therefore:
#
# 0                 -> no penalty (possible nighttime)
# 1–49              -> marginal
# 50–900            -> suitable
# 901–1100          -> marginal
# >1100             -> severe
#

def light_status(value):

    if value == 0:
        return 0

    elif 50 <= value <= 900:
        return 0

    elif (0 < value < 50) or (900 < value <= 1100):
        return 1

    else:
        return 2


final_df["light_status"] = final_df["light_intensity"].apply(
    light_status
)


# ============================================================
# 8. FINAL GREENHOUSE CONDITION
# ============================================================
#
# We avoid a too-simple "one warning = unsuitable" rule.
#
# Suitable:
#     all parameters suitable
#
# Needs Adjustment:
#     1–2 marginal parameters
#     and no severe parameter
#
# Unsuitable:
#     any severe parameter
#     OR 3+ simultaneous marginal parameters
#

STATUS_COLUMNS = [
    "temperature_status",
    "humidity_status",
    "soil_moisture_status",
    "light_status",
    "co2_status"
]


def determine_condition(row):

    values = [row[col] for col in STATUS_COLUMNS]

    severe_count = values.count(2)
    marginal_count = values.count(1)

    if severe_count >= 1:
        return "Unsuitable"

    elif marginal_count >= 3:
        return "Unsuitable"

    elif marginal_count >= 1:
        return "Needs Adjustment"

    else:
        return "Suitable"


final_df["greenhouse_condition"] = final_df.apply(
    determine_condition,
    axis=1
)


# ============================================================
# 9. NUMERIC ML TARGET
# ============================================================

LABEL_MAPPING = {
    "Suitable": 0,
    "Needs Adjustment": 1,
    "Unsuitable": 2
}

final_df["target"] = (
    final_df["greenhouse_condition"]
    .map(LABEL_MAPPING)
)


# ============================================================
# 10. EXACT FARMER-FACING ML FEATURES
# ============================================================

MODEL_FEATURES = [
    "temperature",
    "humidity",
    "soil_moisture",
    "light_intensity",
    "co2"
]

training_df = final_df[
    MODEL_FEATURES +
    [
        "team",
        "datetime",
        "greenhouse_condition",
        "target"
    ]
].copy()


# ============================================================
# 11. VALIDATION
# ============================================================

assert training_df[MODEL_FEATURES].isnull().sum().sum() == 0

assert training_df["target"].isin(
    [0, 1, 2]
).all()

assert len(training_df) == len(final_df)


# ============================================================
# 12. SHOW FINAL CLASS DISTRIBUTION
# ============================================================

print("\n" + "=" * 90)
print("FINAL NAMMA KRUSHI GREENHOUSE DATASET")
print("=" * 90)

print("Rows:", len(training_df))

print("\nML Features:")
for col in MODEL_FEATURES:
    print(" -", col)

print("\nClass distribution:")

class_counts = (
    training_df["greenhouse_condition"]
    .value_counts()
)

print(class_counts)

print("\nClass percentages:")

class_percentages = (
    training_df["greenhouse_condition"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(class_percentages)


# ============================================================
# 13. DISTRIBUTION BY GREENHOUSE TEAM
# ============================================================

print("\nCondition distribution by team (%):")

display(
    pd.crosstab(
        training_df["team"],
        training_df["greenhouse_condition"],
        normalize="index"
    )
    .mul(100)
    .round(2)
)


# ============================================================
# 14. SAVE PERMANENT ARTIFACTS
# ============================================================

OUTPUT_DIR = Path(
    "/kaggle/working/namma_krushi_greenhouse"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# Full final dataset including individual condition statuses
final_df.to_csv(
    OUTPUT_DIR / "greenhouse_final_labeled.csv",
    index=False
)


# Exact dataset used for model training
training_df.to_csv(
    OUTPUT_DIR / "greenhouse_training.csv",
    index=False
)


# Feature configuration
with open(
    OUTPUT_DIR / "feature_columns.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        MODEL_FEATURES,
        f,
        indent=2
    )


# Label configuration
with open(
    OUTPUT_DIR / "label_mapping.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        LABEL_MAPPING,
        f,
        indent=2
    )


# ============================================================
# 15. SAVE LABEL POLICY FOR REPORT / REPRODUCIBILITY
# ============================================================

label_policy = {

    "target":
        "greenhouse_condition",

    "target_origin":
        "Expert-rule-derived tomato greenhouse suitability target",

    "classes": [
        "Suitable",
        "Needs Adjustment",
        "Unsuitable"
    ],

    "features": MODEL_FEATURES,

    "temperature_celsius": {
        "suitable": [18, 27],
        "marginal": [15, 32]
    },

    "relative_humidity_percent": {
        "suitable": [65, 90],
        "marginal": [50, 94]
    },

    "substrate_water_content_percent": {
        "suitable": [68, 88],
        "marginal": [60, 95],
        "note":
            "WC_slab is rockwool/substrate water content, presented to the farmer as root-zone/soil moisture."
    },

    "co2_ppm": {
        "suitable": [400, 1200],
        "marginal": [300, 1500]
    },

    "light_PAR": {
        "zero":
            "No penalty because it may represent nighttime",

        "suitable": [50, 900],

        "marginal_low": [1, 49],

        "marginal_high": [901, 1100],

        "severe_above": 1100
    },

    "overall_condition_rule": {

        "Suitable":
            "All five parameters within suitable ranges",

        "Needs Adjustment":
            "One or two marginal parameters and no severe parameter",

        "Unsuitable":
            "At least one severe parameter or three or more simultaneous marginal parameters"
    },

    "sensor_cleaning": {

        "temperature":
            "(0, 45] C",

        "humidity":
            "[0, 100] %",

        "soil_moisture":
            "(0, 100] %",

        "light":
            "[0, 2000]",

        "co2":
            "[250, 2500] ppm"
    }
}


with open(
    OUTPUT_DIR / "greenhouse_labeling_policy.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        label_policy,
        f,
        indent=2
    )


print("\nSaved final files:")

for p in sorted(OUTPUT_DIR.iterdir()):
    print(" -", p.name)
    
```

```python
# ============================================================
# NAMMA KRUSHI - GREENHOUSE MODEL TRAINING
# ============================================================

import pandas as pd
import numpy as np
import json
import joblib
from pathlib import Path

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    HistGradientBoostingClassifier
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# 1. LOAD FINAL PREPARED DATA
# ============================================================

BASE_DIR = Path("/kaggle/working/namma_krushi_greenhouse")

df_train = pd.read_csv(
    BASE_DIR / "greenhouse_training.csv",
    parse_dates=["datetime"]
)

FEATURES = [
    "temperature",
    "humidity",
    "soil_moisture",
    "light_intensity",
    "co2"
]

LABEL_NAMES = [
    "Suitable",
    "Needs Adjustment",
    "Unsuitable"
]

print("Dataset shape:", df_train.shape)

print("\nTeams:")
print(df_train["team"].value_counts())


# ============================================================
# 2. TEAM-BASED SPLIT
# ============================================================

TEST_TEAM = "Reference"
VALIDATION_TEAM = "IUACAAS"

train_data = df_train[
    ~df_train["team"].isin([
        TEST_TEAM,
        VALIDATION_TEAM
    ])
].copy()

val_data = df_train[
    df_train["team"] == VALIDATION_TEAM
].copy()

test_data = df_train[
    df_train["team"] == TEST_TEAM
].copy()

X_train = train_data[FEATURES]
y_train = train_data["target"]

X_val = val_data[FEATURES]
y_val = val_data["target"]

X_test = test_data[FEATURES]
y_test = test_data["target"]

print("\n" + "=" * 80)
print("DATA SPLIT")
print("=" * 80)

print("Training teams:")
print(sorted(train_data["team"].unique()))

print("\nValidation team:", VALIDATION_TEAM)
print("Test team:", TEST_TEAM)

print("\nTraining rows  :", len(train_data))
print("Validation rows:", len(val_data))
print("Test rows      :", len(test_data))


# ============================================================
# 3. MODELS TO COMPARE
# ============================================================

models = {

    "Logistic Regression": Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            LogisticRegression(
                max_iter=3000,
                class_weight="balanced",
                random_state=42
            )
        )
    ]),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=15,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=250,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        n_jobs=-1,
        random_state=42
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=250,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        n_jobs=-1,
        random_state=42
    ),

    "Hist Gradient Boosting": HistGradientBoostingClassifier(
        max_iter=250,
        learning_rate=0.08,
        max_leaf_nodes=31,
        l2_regularization=0.1,
        random_state=42
    )
}


# ============================================================
# 4. TRAIN AND VALIDATE
# ============================================================

validation_results = []
trained_models = {}

for name, model in models.items():

    print("\n" + "=" * 80)
    print("TRAINING:", name)
    print("=" * 80)

    model.fit(
        X_train,
        y_train
    )

    pred = model.predict(X_val)

    acc = accuracy_score(
        y_val,
        pred
    )

    precision = precision_score(
        y_val,
        pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_val,
        pred,
        average="macro",
        zero_division=0
    )

    macro_f1 = f1_score(
        y_val,
        pred,
        average="macro",
        zero_division=0
    )

    weighted_f1 = f1_score(
        y_val,
        pred,
        average="weighted",
        zero_division=0
    )

    validation_results.append({
        "model": name,
        "accuracy": acc,
        "macro_precision": precision,
        "macro_recall": recall,
        "macro_f1": macro_f1,
        "weighted_f1": weighted_f1
    })

    trained_models[name] = model

    print("Accuracy       :", round(acc, 4))
    print("Macro Precision:", round(precision, 4))
    print("Macro Recall   :", round(recall, 4))
    print("Macro F1       :", round(macro_f1, 4))
    print("Weighted F1    :", round(weighted_f1, 4))


# ============================================================
# 5. MODEL COMPARISON
# ============================================================

results_df = pd.DataFrame(
    validation_results
).sort_values(
    "macro_f1",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 80)
print("VALIDATION MODEL COMPARISON")
print("=" * 80)

display(
    results_df.style.format({
        "accuracy": "{:.4f}",
        "macro_precision": "{:.4f}",
        "macro_recall": "{:.4f}",
        "macro_f1": "{:.4f}",
        "weighted_f1": "{:.4f}"
    })
)


# ============================================================
# 6. SELECT BEST MODEL USING VALIDATION MACRO F1
# ============================================================

best_model_name = results_df.iloc[0]["model"]

print("\nBEST MODEL FROM VALIDATION:")
print(best_model_name)


# ============================================================
# 7. REFIT BEST MODEL ON TRAIN + VALIDATION
# ============================================================

combined_train = pd.concat(
    [
        train_data,
        val_data
    ],
    ignore_index=True
)

X_combined = combined_train[FEATURES]
y_combined = combined_train["target"]

best_model = models[best_model_name]

best_model.fit(
    X_combined,
    y_combined
)


# ============================================================
# 8. FINAL EVALUATION ON NEVER-SEEN REFERENCE GREENHOUSE
# ============================================================

test_pred = best_model.predict(
    X_test
)

test_accuracy = accuracy_score(
    y_test,
    test_pred
)

test_precision = precision_score(
    y_test,
    test_pred,
    average="macro",
    zero_division=0
)

test_recall = recall_score(
    y_test,
    test_pred,
    average="macro",
    zero_division=0
)

test_macro_f1 = f1_score(
    y_test,
    test_pred,
    average="macro",
    zero_division=0
)

test_weighted_f1 = f1_score(
    y_test,
    test_pred,
    average="weighted",
    zero_division=0
)

print("\n" + "=" * 80)
print("FINAL TEST RESULTS - REFERENCE GREENHOUSE")
print("=" * 80)

print("Best model      :", best_model_name)
print("Accuracy        :", round(test_accuracy, 4))
print("Macro Precision :", round(test_precision, 4))
print("Macro Recall    :", round(test_recall, 4))
print("Macro F1        :", round(test_macro_f1, 4))
print("Weighted F1     :", round(test_weighted_f1, 4))


# ============================================================
# 9. CLASSIFICATION REPORT
# ============================================================

report = classification_report(
    y_test,
    test_pred,
    labels=[0, 1, 2],
    target_names=LABEL_NAMES,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(
    report
).transpose()

print("\nCLASSIFICATION REPORT:")

display(
    report_df.round(4)
)


# ============================================================
# 10. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    test_pred,
    labels=[0, 1, 2]
)

cm_df = pd.DataFrame(
    cm,
    index=[
        "Actual_Suitable",
        "Actual_Needs_Adjustment",
        "Actual_Unsuitable"
    ],
    columns=[
        "Predicted_Suitable",
        "Predicted_Needs_Adjustment",
        "Predicted_Unsuitable"
    ]
)

print("\nCONFUSION MATRIX:")

display(cm_df)


# ============================================================
# 11. SAVE EVERYTHING
# ============================================================

MODEL_DIR = BASE_DIR / "model_artifacts"
EVAL_DIR = BASE_DIR / "evaluation"

MODEL_DIR.mkdir(
    parents=True,
    exist_ok=True
)

EVAL_DIR.mkdir(
    parents=True,
    exist_ok=True
)

# Best trained model
joblib.dump(
    best_model,
    MODEL_DIR / "greenhouse_model.pkl"
)

# Model comparison
results_df.to_csv(
    EVAL_DIR / "model_comparison.csv",
    index=False
)

# Classification report
report_df.to_csv(
    EVAL_DIR / "classification_report.csv"
)

# Confusion matrix
cm_df.to_csv(
    EVAL_DIR / "confusion_matrix.csv"
)


# ============================================================
# 12. SAVE TEST PREDICTIONS
# ============================================================

test_results = test_data[
    [
        "team",
        "datetime"
    ] + FEATURES + [
        "greenhouse_condition",
        "target"
    ]
].copy()

test_results["predicted_target"] = test_pred

inverse_mapping = {
    0: "Suitable",
    1: "Needs Adjustment",
    2: "Unsuitable"
}

test_results["predicted_condition"] = (
    test_results["predicted_target"]
    .map(inverse_mapping)
)

test_results["correct"] = (
    test_results["target"] ==
    test_results["predicted_target"]
)

test_results.to_csv(
    EVAL_DIR / "test_predictions.csv",
    index=False
)


# ============================================================
# 13. SAVE MODEL METADATA
# ============================================================

model_metrics = {

    "best_model":
        best_model_name,

    "selection_metric":
        "macro_f1",

    "training_teams":
        sorted(train_data["team"].unique().tolist()),

    "validation_team":
        VALIDATION_TEAM,

    "test_team":
        TEST_TEAM,

    "training_rows":
        int(len(train_data)),

    "validation_rows":
        int(len(val_data)),

    "test_rows":
        int(len(test_data)),

    "features":
        FEATURES,

    "classes":
        LABEL_NAMES,

    "test_accuracy":
        float(test_accuracy),

    "test_macro_precision":
        float(test_precision),

    "test_macro_recall":
        float(test_recall),

    "test_macro_f1":
        float(test_macro_f1),

    "test_weighted_f1":
        float(test_weighted_f1),

    "target_origin":
        "Expert-rule-derived greenhouse suitability target"
}

with open(
    MODEL_DIR / "model_metrics.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        model_metrics,
        f,
        indent=2
    )


# ============================================================
# 14. SAVE FEATURE + CLASS INFORMATION
# ============================================================

with open(
    MODEL_DIR / "feature_columns.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        FEATURES,
        f,
        indent=2
    )

with open(
    MODEL_DIR / "label_mapping.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        {
            "Suitable": 0,
            "Needs Adjustment": 1,
            "Unsuitable": 2
        },
        f,
        indent=2
    )


# ============================================================
# 15. FINAL ARTIFACT LIST
# ============================================================

print("\n" + "=" * 80)
print("MODEL TRAINING COMPLETE")
print("=" * 80)

print("\nModel artifacts:")

for p in sorted(MODEL_DIR.iterdir()):
    print(" -", p.name)

print("\nEvaluation artifacts:")

for p in sorted(EVAL_DIR.iterdir()):
    print(" -", p.name)
```

```python
# ============================================================
# NAMMA KRUSHI - GREENHOUSE XAI + FARMER INPUT TEST
# ============================================================

import pandas as pd
import numpy as np
import json
import joblib
from pathlib import Path
import matplotlib.pyplot as plt

from sklearn.inspection import permutation_importance

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

BASE_DIR = Path("/kaggle/working/namma_krushi_greenhouse")
MODEL_DIR = BASE_DIR / "model_artifacts"
EVAL_DIR = BASE_DIR / "evaluation"
XAI_DIR = BASE_DIR / "xai"

XAI_DIR.mkdir(parents=True, exist_ok=True)

FEATURES = [
    "temperature",
    "humidity",
    "soil_moisture",
    "light_intensity",
    "co2"
]

LABELS = {
    0: "Suitable",
    1: "Needs Adjustment",
    2: "Unsuitable"
}

# ------------------------------------------------------------
# 2. LOAD SAVED MODEL
# ------------------------------------------------------------

model = joblib.load(
    MODEL_DIR / "greenhouse_model.pkl"
)

print("Loaded model:")
print(type(model).__name__)


# ============================================================
# 3. RANDOM FOREST BUILT-IN FEATURE IMPORTANCE
# ============================================================

feature_importance_df = pd.DataFrame({
    "feature": FEATURES,
    "importance": model.feature_importances_
}).sort_values(
    "importance",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 80)
print("GLOBAL FEATURE IMPORTANCE")
print("=" * 80)

display(feature_importance_df)

feature_importance_df.to_csv(
    XAI_DIR / "feature_importance.csv",
    index=False
)

plt.figure(figsize=(8, 5))
plt.bar(
    feature_importance_df["feature"],
    feature_importance_df["importance"]
)

plt.title("Greenhouse Model Feature Importance")
plt.xlabel("Feature")
plt.ylabel("Importance")
plt.xticks(rotation=30)
plt.tight_layout()

plt.savefig(
    XAI_DIR / "feature_importance.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 4. PERMUTATION IMPORTANCE
# ============================================================
# Use sample to keep execution fast.

sample_size = min(10000, len(X_test))

sample_idx = X_test.sample(
    n=sample_size,
    random_state=42
).index

X_perm = X_test.loc[sample_idx]
y_perm = y_test.loc[sample_idx]

perm = permutation_importance(
    model,
    X_perm,
    y_perm,
    n_repeats=5,
    random_state=42,
    scoring="f1_macro",
    n_jobs=-1
)

permutation_df = pd.DataFrame({
    "feature": FEATURES,
    "importance_mean": perm.importances_mean,
    "importance_std": perm.importances_std
}).sort_values(
    "importance_mean",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 80)
print("PERMUTATION IMPORTANCE")
print("=" * 80)

display(permutation_df)

permutation_df.to_csv(
    XAI_DIR / "permutation_importance.csv",
    index=False
)


# ============================================================
# 5. SHAP XAI
# ============================================================

try:
    import shap

    print("\nCalculating SHAP values...")

    X_shap = X_test.sample(
        n=min(1000, len(X_test)),
        random_state=42
    )

    explainer = shap.TreeExplainer(model)

    shap_values = explainer.shap_values(X_shap)

    # Handle old/new SHAP output formats
    if isinstance(shap_values, list):

        mean_abs_shap = np.mean(
            [
                np.abs(class_values).mean(axis=0)
                for class_values in shap_values
            ],
            axis=0
        )

    else:

        shap_array = np.asarray(shap_values)

        if shap_array.ndim == 3:
            # samples x features x classes
            mean_abs_shap = np.abs(
                shap_array
            ).mean(axis=(0, 2))

        elif shap_array.ndim == 2:
            mean_abs_shap = np.abs(
                shap_array
            ).mean(axis=0)

        else:
            raise ValueError(
                f"Unexpected SHAP shape: {shap_array.shape}"
            )

    shap_importance_df = pd.DataFrame({
        "feature": FEATURES,
        "mean_abs_shap": mean_abs_shap
    }).sort_values(
        "mean_abs_shap",
        ascending=False
    ).reset_index(drop=True)

    print("\n" + "=" * 80)
    print("SHAP GLOBAL IMPORTANCE")
    print("=" * 80)

    display(shap_importance_df)

    shap_importance_df.to_csv(
        XAI_DIR / "shap_global_importance.csv",
        index=False
    )

    print("\nSHAP completed successfully.")

except Exception as e:

    print("\nSHAP could not run:")
    print(e)

    print(
        "\nBuilt-in Random Forest importance and "
        "permutation importance are still available."
    )


# ============================================================
# 6. FARMER INPUT PREDICTION FUNCTION
# ============================================================

def predict_greenhouse(
    temperature,
    humidity,
    soil_moisture,
    light_intensity,
    co2
):

    input_df = pd.DataFrame([{
        "temperature": temperature,
        "humidity": humidity,
        "soil_moisture": soil_moisture,
        "light_intensity": light_intensity,
        "co2": co2
    }])

    prediction = int(
        model.predict(input_df)[0]
    )

    probabilities = model.predict_proba(
        input_df
    )[0]

    result = {
        "prediction": LABELS[prediction],

        "confidence": float(
            probabilities[prediction]
        ),

        "probabilities": {
            LABELS[int(cls)]: float(prob)
            for cls, prob in zip(
                model.classes_,
                probabilities
            )
        }
    }

    return result


# ============================================================
# 7. TEST REALISTIC FARMER INPUTS
# ============================================================

test_cases = [

    {
        "name": "Good greenhouse conditions",
        "temperature": 23,
        "humidity": 80,
        "soil_moisture": 78,
        "light_intensity": 250,
        "co2": 750
    },

    {
        "name": "Conditions needing adjustment",
        "temperature": 29,
        "humidity": 92,
        "soil_moisture": 76,
        "light_intensity": 250,
        "co2": 700
    },

    {
        "name": "Clearly unsuitable temperature",
        "temperature": 34,
        "humidity": 80,
        "soil_moisture": 76,
        "light_intensity": 250,
        "co2": 700
    },

    {
        "name": "Low root-zone moisture",
        "temperature": 23,
        "humidity": 80,
        "soil_moisture": 55,
        "light_intensity": 250,
        "co2": 700
    },

    {
        "name": "Very high humidity",
        "temperature": 23,
        "humidity": 97,
        "soil_moisture": 76,
        "light_intensity": 250,
        "co2": 700
    }
]


farmer_results = []

print("\n" + "=" * 80)
print("FARMER INPUT TESTS")
print("=" * 80)

for case in test_cases:

    result = predict_greenhouse(
        case["temperature"],
        case["humidity"],
        case["soil_moisture"],
        case["light_intensity"],
        case["co2"]
    )

    row = {
        **case,
        "prediction": result["prediction"],
        "confidence": round(
            result["confidence"],
            4
        )
    }

    farmer_results.append(row)

    print("\nCASE:", case["name"])

    print(
        "Temperature:",
        case["temperature"]
    )

    print(
        "Humidity:",
        case["humidity"]
    )

    print(
        "Soil Moisture:",
        case["soil_moisture"]
    )

    print(
        "Light:",
        case["light_intensity"]
    )

    print(
        "CO2:",
        case["co2"]
    )

    print(
        "Prediction:",
        result["prediction"]
    )

    print(
        "Confidence:",
        round(
            result["confidence"] * 100,
            2
        ),
        "%"
    )

    print(
        "Class probabilities:",
        {
            k: round(v, 4)
            for k, v in
            result["probabilities"].items()
        }
    )


# ============================================================
# 8. SAVE FARMER TEST RESULTS
# ============================================================

farmer_results_df = pd.DataFrame(
    farmer_results
)

farmer_results_df.to_csv(
    XAI_DIR / "farmer_input_tests.csv",
    index=False
)

print("\n" + "=" * 80)
print("XAI ARTIFACTS CREATED")
print("=" * 80)

for p in sorted(XAI_DIR.iterdir()):
    print(" -", p.name)
```
